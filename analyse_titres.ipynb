```python
import pandas as pd
import json
# uploade d’abord votre JSON
with open('Openopus_works.json', 'r', encoding='utf-8') as f:
    works = json.load(f)

# Si vous voulez analyser les œuvres
df_works = pd.json_normalize(works['composers'])
display(df_works.head())
```

```python
import spacy
import networkx as nx

import matplotlib.pyplot as plt

# charger un modèle français pour NER et POS
nlp = spacy.load("fr_core_news_sm")

# construire le dictionnaire compositeur → liste de titres
composer_works = {
    comp['name']: [w['title'] for w in comp['works']]
    for comp in works['composers']
}

# créer le graphe
G = nx.Graph()
i=0
for composer, titles in composer_works.items():
    i+=1
    print(f"Composer {i} sur 220: {composer}")
    G.add_node(composer, type='composer')
    for title in titles:
        doc = nlp(title)
        # entités nommées
        for ent in doc.ents:
            term = ent.text.lower()
            G.add_node(term, type='term')
            G.add_edge(composer, term)
        # noms communs
        for token in doc:
            if token.pos_ == "NOUN":
                term = token.text.lower()
                G.add_node(term, type='term')
                G.add_edge(composer, term)
```
